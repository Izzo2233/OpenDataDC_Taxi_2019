```python
def random_sample_file(file, n=10000, chunksize=100000):
    samples = []

    for chunk in pd.read_csv(file, chunksize=chunksize):
        samples.append(
            chunk.sample(
                n=min(n, len(chunk)),
                random_state=42
            )
        )

    data = pd.concat(samples, ignore_index=True)

    return data.sample(
        n=min(n, len(data)),
        random_state=42
    )
```
